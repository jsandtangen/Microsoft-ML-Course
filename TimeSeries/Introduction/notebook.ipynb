```python
import importlib.util
import os
from pathlib import Path
import matplotlib.pyplot as plt

utils_path = None
for path in [Path.cwd(), *Path.cwd().parents]:
    if (path / 'common' / 'utils.py').exists():
        utils_path = path / 'common' / 'utils.py'
        break
    if (path / 'TimeSeries' / 'common' / 'utils.py').exists():
        utils_path = path / 'TimeSeries' / 'common' / 'utils.py'
        break

if utils_path is None:
    raise FileNotFoundError(f"Could not find TimeSeries/common/utils.py from {Path.cwd()}")

spec = importlib.util.spec_from_file_location('timeseries_common_utils', utils_path)
utils = importlib.util.module_from_spec(spec)
spec.loader.exec_module(utils)
load_data = utils.load_data
%matplotlib inline
```

```python
data_dir = utils_path.parent.parent / 'data'
```
