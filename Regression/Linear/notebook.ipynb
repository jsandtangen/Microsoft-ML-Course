```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from datetime import datetime
import matplotlib.pyplot as plt
```

```python
pumpkins = pd.read_csv("../data/US-pumpkins.csv")
pumpkins.head()
```

```python
# case=True betyr at søket skiller mellom store og små bokstaver.
# regex=True betyr at søket kan bruke mønstre, f.eks. "bushel|carton" for bushel eller carton.
pumpkins = pumpkins[pumpkins['Package'].str.contains('bushel', case=True, regex=True)]

columns_to_select = ['Package', 'Variety', 'City Name', 'Low Price', 'High Price', 'Date']
pumpkins = pumpkins.loc[:, columns_to_select]

price = (pumpkins['Low Price'] + pumpkins['High Price']) / 2

month = pd.DatetimeIndex(pumpkins['Date']).month
day_of_year = pd.to_datetime(pumpkins['Date']).apply(lambda dt: (dt-datetime(dt.year,1,1)).days)

new_pumpkins = pd.DataFrame(
    {'Month': month, 
     'DayOfYear' : day_of_year, 
     'Variety': pumpkins['Variety'], 
     'City': pumpkins['City Name'], 
     'Package': pumpkins['Package'], 
     'Low Price': pumpkins['Low Price'],
     'High Price': pumpkins['High Price'], 
     'Price': price})

new_pumpkins.loc[new_pumpkins['Package'].str.contains('1 1/9'), 'Price'] = price/1.1
new_pumpkins.loc[new_pumpkins['Package'].str.contains('1/2'), 'Price'] = price*2

new_pumpkins.head()
```

```python
plt.scatter('Month','Price',data=new_pumpkins)
plt.xlabel('Month')
plt.ylabel('Price')
```

```python
plt.scatter('DayOfYear','Price',data=new_pumpkins)
plt.xlabel('Day of Year')
plt.ylabel('Price')
```

```python
print(new_pumpkins['Month'].corr(new_pumpkins['Price']))
print(new_pumpkins['DayOfYear'].corr(new_pumpkins['Price']))
```

```python
ax=None
colors = ['red','blue','green','yellow']
for i,var in enumerate(new_pumpkins['Variety'].unique()):
    df = new_pumpkins[new_pumpkins['Variety']==var]
    ax = df.plot.scatter('DayOfYear','Price',ax=ax,c=colors[i],label=var)

ax.set_xlabel('Day of Year')
ax.legend(fontsize=7, markerscale=0.5)
```

```python
new_pumpkins.groupby('Variety')['Price'].mean().plot(kind='bar')
plt.ylabel('Average Price')
```

```python
pie_pumpkins = new_pumpkins[new_pumpkins['Variety']=='PIE TYPE']
pie_pumpkins.plot.scatter('DayOfYear','Price') 

plt.xlabel('Day of Year')
```

```python
pie_pumpkins.dropna(inplace=True)
pie_pumpkins.info()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
```

```python
X = pie_pumpkins['DayOfYear'].to_numpy().reshape(-1,1)
y = pie_pumpkins['Price']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
lin_reg = LinearRegression()
lin_reg.fit(X_train,y_train)
```

```python
pred = lin_reg.predict(X_test)

mse = np.sqrt(mean_squared_error(y_test,pred))
print(f'Mean error: {mse:3.3} ({mse/np.mean(pred)*100:3.3}%)')
```

```python
score = lin_reg.score(X_train,y_train)
print('Model determination: ', score)
```

```python
plt.scatter(X_test,y_test)
plt.plot(X_test,pred)
```

Modellen har en gjennomsnittlig feil på rundt 2 poeng, som tilsvarer omtrent 17 %. Dette viser at modellen bommer en del på prediksjonene.

R²-verdien er også lav, cirka 0,06, som betyr at modellen bare forklarer omtrent 6 % av variasjonen i dataene. En R²-verdi nær 0 betyr at modellen nesten ikke bruker input-dataene til å forklare resultatet, og at den ikke er mye bedre enn å bare predikere gjennomsnittet.

Samlet sett tyder dette på at modellen har svak prediksjonsevne, og at den bør forbedres med bedre variabler, mer data, feature engineering eller en annen modell.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline

pipeline = make_pipeline(PolynomialFeatures(2), LinearRegression())

pipeline.fit(X_train,y_train)
```

```python
pred = pipeline.predict(X_test)

mse = np.sqrt(mean_squared_error(y_test,pred))
print(f'Mean error: {mse:3.3} ({mse/np.mean(pred)*100:3.3}%)')

score = pipeline.score(X_train,y_train)
print('Model determination: ', score)
```

```python
X_range = np.linspace(X_test.min(), X_test.max(), 100).reshape(-1,1)
y_range = pipeline.predict(X_range)

plt.scatter(X_test, y_test)
plt.plot(X_range, y_range)
```

```python
X = pd.get_dummies(new_pumpkins['Variety']) \
        .join(new_pumpkins['Month']) \
        .join(pd.get_dummies(new_pumpkins['City'])) \
        .join(pd.get_dummies(new_pumpkins['Package']))
y = new_pumpkins['Price']

# make train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# setup and train the pipeline
pipeline = make_pipeline(PolynomialFeatures(2), LinearRegression())
pipeline.fit(X_train,y_train)

# predict results for test data
pred = pipeline.predict(X_test)

# calculate MSE and determination
mse = np.sqrt(mean_squared_error(y_test,pred))
print(f'Mean error: {mse:3.3} ({mse/np.mean(pred)*100:3.3}%)')

score = pipeline.score(X_train,y_train)
print('Model determination: ', score)
```
